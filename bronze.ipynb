```python
last_load_date = "1900-01-01"
```

```python
if spark.catalog.tableExists("data_modeling.bronze.customers_bronze"):

    last_load_date= spark.sql("select max(date_joined) from data_modeling.bronze.customers_bronze")\
                   .collect()[0][0]
else:
    last_load_date= "1900-01-01"
```

```python
last_load_date
```

```python
spark.sql(f"""select * from data_modeling.default.customers where date_joined > date('{last_load_date}') """).createOrReplaceTempView("customers_source")
```

```python


%sql
CREATE OR REPLACE TABLE data_modeling.bronze.customers_bronze
AS 
SELECT * FROM customers_source
```
