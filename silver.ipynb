```python
spark.sql     ("""
               select *,
                    upper(first_name) as name,
                    upper(last_name) as surname,
                    date (current_timestamp()) as process_date   
                 from 
                     data_modeling.bronze.customers_bronze 
                     
                     """).createOrReplaceTempView("customers_silverSource")
```

```python
%sql
select * from customers_silverSource
```

```python
%sql
create table if not exists data_modeling.silver.customers_silver
              as
              select * from customers_silverSource
```

```python
%sql
merge into data_modeling.silver.customers_silver
using customers_silverSource
on data_modeling.silver.customers_silver.customer_id = customers_silverSource.customer_id
when matched then update set *
when not matched then insert *
```
